## Looking For Common Threads Across Successful Seasons
This notebook will aim to answer my first research question and identify common elements across successful seasons throughout the history of the Houston Texans

```python
# Import libraries
import pandas as pd
from sqlalchemy import create_engine
import os
from dotenv import load_dotenv
```

```python
# Connect to PostgreSQL database
load_dotenv()

DB_NAME = os.getenv('DB_NAME')
DB_USER = os.getenv('DB_USER')
DB_PASSWORD = os.getenv('DB_PASSWORD')
DB_PORT = os.getenv('DB_PORT')
DB_HOST = os.getenv('DB_HOST')

engine = create_engine(f'postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}')
```

```python
# Look at summary of all Texans seasons
query = """SELECT * FROM texans_season_summary"""
season_summary = pd.read_sql(query, engine)
season_summary
```

```python
# Correlations against win_pct
metrics = ['turnover_margin', 'redzone_td_pct_differential', 'epa_differential']
season_summary[metrics + ['win_pct']].corr()['win_pct'].drop('win_pct')
```

It makes sense that EPA differential is the strongest single predictor of win_pct as EPA differential is basically saying "how much better were you than your opponent".

Turnover margin also shows a strong correlation with win_pct, which indicates that a strong defense and safe offense could lead to success.

```python
# Comparison against made_playoffs
season_summary.groupby('made_playoffs')[metrics].mean()
```

Unsurprisingly, turnover margin has been high on years that the Texans have made the playoffs and negative in years where they did not. Also, as a longtime fan, it isn't surprising that the redzone td pct differential isn't too different between playoff and non-playoff seasons.
